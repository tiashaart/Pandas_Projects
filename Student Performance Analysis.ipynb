```python
#data cleaning

df.isnull().sum()
```

```python
#handles missing values

df.fillna(df.mean(), inplace=True)
df.drop_duplicates(inplace=True)
```

```python
#Basic Analysis
# 1. Number of students
print("Total students:", len(df))

# 2. Average GPA
print("Average GPA:", df["GPA"].mean())

# 3. Highest and Lowest GPA
print("Highest GPA:", df["GPA"].max())
print("Lowest GPA:", df["GPA"].min())

# 4. Unique GradeClass values
print("Unique GradeClass count:", df["GradeClass"].nunique())

# 5. Average StudyTimeWeekly
print("Average Study Time Weekly:", df["StudyTimeWeekly"].mean())
```

```python
#6. Distribution of GPA

df["GPA"].hist()
df["GPA"].describe()
```

```python
#11. StudyTime vs GPA

df[["StudyTimeWeekly", "GPA"]].corr()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")

plt.figure(figsize=(6,4))
sns.histplot(df["GPA"], bins=20, kde=True)
plt.title("Distribution of GPA")
plt.show()
```

```python

plt.figure(figsize=(6,4))
sns.scatterplot(x="StudyTimeWeekly", y="GPA", data=df)
plt.title("Study Time vs GPA")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.histplot(df["StudyTimeWeekly"], bins=20, kde=True)
plt.title("Distribution of Study Time Weekly")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x="GradeClass", y="GPA", data=df)
plt.title("GPA by Grade Class")
plt.show()
```

```python
plt.figure(figsize=(6,4))
df.groupby("GradeClass")["GPA"].mean().plot(kind="bar")
plt.title("Average GPA per Grade Class")
plt.ylabel("GPA")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.barplot(x="GradeClass", y="StudyTimeWeekly", data=df)
plt.title("Study Time by Grade Class")
plt.show()
```

```python
sns.pairplot(df[["GPA", "StudyTimeWeekly", "Absences"]])
plt.show()
```
