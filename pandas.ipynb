```python
df = pd.read_csv('archive (1).zip')
z_scaled = df.copy()
for column in z_scaled.select_dtypes(include='number').columns:
    z_scaled[column] = (z_scaled[column] - z_scaled[column].mean()) / z_scaled[column].std()
print(z_scaled)
z_scaled.plot(kind="hist")
plt.show()
```
